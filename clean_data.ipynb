## 1. Load the files


```python
import pandas as pd

customers = pd.read_csv('customers_master.csv')
products = pd.read_csv('products_master.csv')
orders = pd.read_csv('orders.csv')
batch = pd.read_csv('new_daily_batch.csv')
summary = pd.read_csv('previous_days_summary.csv')
print(customers.shape, products.shape, orders.shape, batch.shape, summary.shape)
```

## 2. Explore: what is wrong?


```python
print(orders.isna().sum())
print('exact duplicate rows in orders:', orders.duplicated().sum())
print('regions before cleaning:', customers['Region'].unique())
print('categories before cleaning:', products['Category'].unique())
print('status before cleaning:', orders['Order_Status'].unique())
print('batch quantity values:', batch['Quantity'].unique())
print('batch discount values:', batch['Discount'].unique())
print('batch payment values:', batch['Payment_Method'].unique())
```

## 3. Clean the master files

Strip spaces and standardise case. Regions go from 13 raw spellings to the 4 real regions, product categories from 8 spellings to 5.


```python
customers['Region'] = customers['Region'].str.strip().str.title()
customers['Customer_Segment'] = customers['Customer_Segment'].str.strip().str.title()
products['Category'] = products['Category'].str.strip().str.title()
print(sorted(customers['Region'].unique()))
print(sorted(products['Category'].unique()))
```

## 4. Clean the order files

One order can have several lines (one per product), so repeated `Order_ID` is normal - only exact duplicate rows are dropped. Missing `Customer_ID` / invalid `Order_Date` are recovered from the other lines of the same order. Invalid rows are saved to `rejected/` with a reason instead of being silently deleted.
